# Kalpitiya Deforestation Analysis
### Landsat 8 & 9 — Surface Reflectance — 2016 to 2025

This notebook runs a full deforestation analysis using USGS Landsat Collection 2 Level-2 data. The scenes here are large (~7700 x 7600 px), so the notebook is built around one strict rule: **only one scene lives in RAM at a time**. Every intermediate result is saved to disk and reloaded on demand.

**Six passes, in order:**

| Pass | What happens | Output |
|---|---|---|
| 1 | Cloud-mask + scale + save stacked TIFs | `outputs/stacked/` |
| 2 | Compute NDVI, NDWI, NDBI, NBR, EVI | `outputs/ndvi/` etc. |
| 3 | Train Random Forest classifier | `outputs/models/` |
| 4 | Classify all scenes | `outputs/classmaps/` |
| 5 | Change detection | `outputs/change/` |
| 6 | Trend analysis and forecast to 2030 | `outputs/stats/` |

Restart-safe: if you stop and restart the kernel, already-processed TIFs are detected and skipped automatically.

```python
# ── Only change these two lines ──────────────────────────────────────────────
RAW_DIR    = "data/raw"
OUT_DIR    = "out"
DOWNSAMPLE = 2       # 1=full res (~1.6GB/scene), 2=half res (~415MB), 4=quarter

# Landsat Collection 2 Level-2 scaling:  SR = DN * 0.0000275 + (-0.2)
SCALE  = 0.0000275
OFFSET = -0.2

# QA_PIXEL bit flags (from the Landsat C2 product guide)
QA_FILL   = 0x0001
QA_DILCLD = 0x0002
QA_CIRRUS = 0x0004
QA_CLOUD  = 0x0008
QA_SHADOW = 0x0010

# Land cover classes
CLASS_NAMES  = {0: "Water", 1: "Vegetation", 2: "Bare Soil", 3: "Built-Up"}
CLASS_COLORS = {-1: "#cccccc", 0: "#2171b5", 1: "#238b45", 2: "#d4a843", 3: "#cb181d"}

VEG_THRESHOLD = 0.25

# Make sure DOWNSAMPLE is always a plain int — prevents tuple arithmetic errors later
DOWNSAMPLE = int(DOWNSAMPLE)
PX_HA = 0.09 * (DOWNSAMPLE ** 2)   # hectares per pixel at chosen resolution

for sub in ["stacked","ndvi","ndwi","ndbi","nbr","evi",
            "classmaps","change","graphs","stats","models"]:
    os.makedirs(os.path.join(OUT_DIR, sub), exist_ok=True)

print(f"RAW_DIR    : {os.path.abspath(RAW_DIR)}")
print(f"OUT_DIR    : {os.path.abspath(OUT_DIR)}")
print(f"DOWNSAMPLE : {DOWNSAMPLE}x  →  ~{7761//DOWNSAMPLE} x {7611//DOWNSAMPLE} px per scene")
print(f"PX_HA      : {PX_HA} ha per pixel")
```

```python
def parse_landsat_name(fname):
    base = os.path.basename(fname)
    # Pattern: LC08_L2SP_142054_20160203_20200907_02_T1_SR_B4.TIF
    m = re.match(
        r"(LC0[89])_L2SP_(\d{6})_(\d{8})_\d{8}_\d{2}_T\d_"
        r"(SR_B[1-7]|QA_PIXEL|QA_RADSAT|SR_QA_AEROSOL)",
        base, re.IGNORECASE
    )
    if not m:
        return None
    acq = m.group(3)
    return {
        "sensor":   m.group(1).upper(),
        "path_row": m.group(2),
        "date":     datetime.strptime(acq, "%Y%m%d"),
        "band":     m.group(4).upper(),
        "label":    f"{acq[:4]}_{acq[4:6]}",
    }


def discover_scenes(raw_dir):
    raw_path = Path(raw_dir)
    if not raw_path.exists():
        raise FileNotFoundError(f"Not found: {raw_dir}")
    scenes = {}
    for folder in sorted(raw_path.iterdir()):
        if not folder.is_dir():
            continue
        label = folder.name
        tifs  = sorted(list(folder.glob("*.TIF")) + list(folder.glob("*.tif")))
        if not tifs:
            print(f"  {label}: no TIF files — skipping")
            continue
        entry = {}
        for tif in tifs:
            info = parse_landsat_name(tif.name)
            if info:
                entry[info["band"]] = str(tif)
                if "_meta" not in entry:
                    entry["_meta"] = {k: v for k, v in info.items() if k != "band"}
        if entry:
            sr  = sum(1 for k in entry if k.startswith("SR_B"))
            has = "QA_PIXEL" in entry
            dt  = entry["_meta"]["date"].strftime("%Y-%m-%d")
            print(f"  {label}  {entry['_meta']['sensor']}  {dt}  {sr}/7 bands  QA:{'yes' if has else 'NO'}")
            scenes[label] = entry
    print(f"\n  Total: {len(scenes)} scenes found")
    return scenes


print(f"Scanning {os.path.abspath(RAW_DIR)} ...\n")
SCENES = discover_scenes(RAW_DIR)
LABELS = sorted(SCENES.keys())
```

```python
def load_band(path, downsample=1):
    """Load one SR band, scale to surface reflectance, return float32 array."""
    with rasterio.open(path) as src:
        if downsample == 1:
            arr     = src.read(1).astype(np.float32)
            profile = src.profile.copy()
        else:
            out_h = src.height // downsample
            out_w = src.width  // downsample
            arr   = src.read(
                1,
                out_shape=(out_h, out_w),
                resampling=Resampling.bilinear
            ).astype(np.float32)
            new_t   = src.transform * src.transform.scale(
                src.width  / out_w,
                src.height / out_h
            )
            profile = src.profile.copy()
            profile.update(height=out_h, width=out_w, transform=new_t)
        nodata = src.nodata
    if nodata is not None:
        arr[arr == nodata] = np.nan
    arr = np.clip(arr * SCALE + OFFSET, 0.0, 1.0)
    return arr, profile


def load_qa(path, downsample=1):
    """Load QA_PIXEL, return boolean mask — True = bad pixel."""
    with rasterio.open(path) as src:
        if downsample == 1:
            qa = src.read(1).astype(np.uint16)
        else:
            out_h = src.height // downsample
            out_w = src.width  // downsample
            qa    = src.read(
                1,
                out_shape=(out_h, out_w),
                resampling=Resampling.nearest
            ).astype(np.uint16)
    bad = (((qa & QA_FILL)   > 0) |
           ((qa & QA_CLOUD)  > 0) |
           ((qa & QA_DILCLD) > 0) |
           ((qa & QA_SHADOW) > 0) |
           ((qa & QA_CIRRUS) > 0))
    return bad


def save_raster(arr, profile, path):
    """Save a numpy array as a compressed GeoTIFF."""
    os.makedirs(os.path.dirname(path) or ".", exist_ok=True)
    if arr.ndim == 2:
        arr = arr[np.newaxis]
    p = profile.copy()
    p.update(count=arr.shape[0], dtype="float32",
             driver="GTiff", compress="lzw", nodata=float("nan"),
             BIGTIFF="YES")
    with rasterio.open(path, "w", **p) as dst:
        dst.write(arr.astype(np.float32))


def pct_stretch(arr, lo=2, hi=98):
    """Percentile contrast stretch for display — maps lo..hi percentiles to 0..1."""
    p_lo = np.nanpercentile(arr, lo)
    p_hi = np.nanpercentile(arr, hi)
    return np.clip((arr - p_lo) / (p_hi - p_lo + 1e-9), 0.0, 1.0)


def read_index_tif(label, name):
    """Load a saved single-band index TIF from disk."""
    path = f"{OUT_DIR}/{name}/{label}_{name}.tif"
    with rasterio.open(path) as src:
        return src.read(1).astype(np.float32)


print("I/O helpers defined.")
```

## 3. Scene Discovery

Builds a catalogue of file paths from the folder names. Nothing is loaded into memory here.

```python
# Pass 1 — load, cloud-mask, scale, save stacked TIF for every scene.
# Only one scene lives in RAM at a time. After save() we del and gc.collect().
# If the stacked TIF already exists the scene is skipped automatically.

print(f"Preprocessing  (DOWNSAMPLE={DOWNSAMPLE}) ...\n")
cloud_log = {}

for label in LABELS:
    out_path  = f"{OUT_DIR}/stacked/{label}_stack.tif"
    side_path = f"{OUT_DIR}/stacked/{label}_cloud.txt"

    if os.path.exists(out_path):
        cloud_log[label] = float(open(side_path).read()) if os.path.exists(side_path) else 0.0
        print(f"  {label}  already done  (cloud={cloud_log[label]:.1f}%)")
        continue

    t0    = time.time()
    scene = SCENES[label]
    bands   = []
    profile = None

    for i in range(1, 8):
        bk = f"SR_B{i}"
        if bk not in scene:
            print(f"  {label}  WARNING: {bk} missing — scene skipped")
            bands = []
            break
        arr, prof = load_band(scene[bk], downsample=DOWNSAMPLE)
        bands.append(arr)
        del arr
        if profile is None:
            profile = prof

    if not bands:
        continue

    stack = np.stack(bands, axis=0).astype(np.float32)   # (7, H, W)
    del bands
    gc.collect()

    cloud_pct = 0.0
    if "QA_PIXEL" in scene:
        bad_px    = load_qa(scene["QA_PIXEL"], downsample=DOWNSAMPLE)
        cloud_pct = 100.0 * float(bad_px.sum()) / float(bad_px.size)
        stack[:, bad_px] = np.nan
        del bad_px

    save_raster(stack, profile, out_path)
    cloud_log[label] = cloud_pct
    with open(side_path, "w") as f:
        f.write(str(cloud_pct))

    h, w   = stack.shape[1], stack.shape[2]
    mem_mb = stack.nbytes / 1e6
    del stack
    gc.collect()

    print(f"  {label}  {h}x{w}  cloud={cloud_pct:.1f}%  {mem_mb:.0f}MB  ({time.time()-t0:.1f}s)")

print(f"\nDone. Stacked TIFs in {OUT_DIR}/stacked/")
```

```python
def safe_ratio(a, b):
    """Normalised difference (a-b)/(a+b), NaN where denominator is near zero."""
    a = a.astype(np.float32)
    b = b.astype(np.float32)
    with np.errstate(invalid="ignore", divide="ignore"):
        denom  = a + b
        result = np.where(np.abs(denom) > 1e-9, (a - b) / denom, np.nan)
    return result.astype(np.float32)


def compute_indices(stack):
    """
    Compute spectral indices from a 7-band Landsat stack.

    Band order (0-based index):
      0=B1  1=B2(Blue)  2=B3(Green)  3=B4(Red)
      4=B5(NIR)  5=B6(SWIR1)  6=B7(SWIR2)
    """
    blue  = stack[1];  green = stack[2];  red   = stack[3]
    nir   = stack[4];  swir1 = stack[5];  swir2 = stack[6]

    ndvi = np.clip(safe_ratio(nir,   red),   -1.0, 1.0)
    ndwi = np.clip(safe_ratio(green, nir),   -1.0, 1.0)
    ndbi = np.clip(safe_ratio(swir1, nir),   -1.0, 1.0)
    nbr  = np.clip(safe_ratio(nir,   swir2), -1.0, 1.0)

    # EVI has a different denominator
    with np.errstate(invalid="ignore", divide="ignore"):
        d_evi = nir.astype(np.float32) + 6.0 * red.astype(np.float32) \
                - 7.5 * blue.astype(np.float32) + 1.0
        evi = np.where(
            np.abs(d_evi) > 1e-9,
            2.5 * (nir.astype(np.float32) - red.astype(np.float32)) / d_evi,
            np.nan
        ).astype(np.float32)
    evi = np.clip(evi, -1.0, 1.0)

    return {"ndvi": ndvi, "ndwi": ndwi, "ndbi": ndbi, "nbr": nbr, "evi": evi}


print("Index functions defined.")
```

```python
# Pass 2 — compute all spectral indices, save each to its own TIF, free RAM.
# We also collect NDVI statistics here into a tiny DataFrame that stays in RAM.

print("Computing spectral indices ...\n")
stats_rows = []

for label in LABELS:
    stk_path = f"{OUT_DIR}/stacked/{label}_stack.tif"
    if not os.path.exists(stk_path):
        print(f"  {label}: stacked TIF missing — run preprocessing cell first")
        continue

    t0 = time.time()

    with rasterio.open(stk_path) as src:
        stack   = src.read().astype(np.float32)   # (7, H, W)
        profile = src.profile.copy()

    idx = compute_indices(stack)
    del stack
    gc.collect()

    # Save each index TIF — only if not already saved
    one_band = profile.copy()
    one_band.update(count=1, dtype="float32", compress="lzw",
                    nodata=float("nan"), BIGTIFF="YES")
    for iname, arr in idx.items():
        out_path = f"{OUT_DIR}/{iname}/{label}_{iname}.tif"
        if not os.path.exists(out_path):
            save_raster(arr, one_band, out_path)

    # Collect stats while arrays are in memory (cheap — just numbers)
    valid = idx["ndvi"][~np.isnan(idx["ndvi"])].ravel()
    if len(valid) > 1000:
        yr, mo = int(label[:4]), int(label[5:])
        stats_rows.append({
            "label":        label,
            "year":         yr,
            "month":        mo,
            "date_frac":    yr + (mo - 1) / 12.0,
            "mean_ndvi":    float(valid.mean()),
            "std_ndvi":     float(valid.std()),
            "median_ndvi":  float(np.median(valid)),
            "p10":          float(np.percentile(valid, 10)),
            "p90":          float(np.percentile(valid, 90)),
            "mean_ndwi":    float(np.nanmean(idx["ndwi"])),
            "mean_ndbi":    float(np.nanmean(idx["ndbi"])),
            "pct_dense":    float(100.0 * (valid > 0.50).mean()),
            "pct_moderate": float(100.0 * ((valid > 0.25) & (valid <= 0.50)).mean()),
            "pct_sparse":   float(100.0 * ((valid > 0.00) & (valid <= 0.25)).mean()),
            "pct_nonveg":   float(100.0 * (valid <= 0.00).mean()),
            "n_valid_px":   int(len(valid)),
        })

    mn = float(np.nanmean(idx["ndvi"]))
    mw = float(np.nanmean(idx["ndwi"]))
    mb = float(np.nanmean(idx["ndbi"]))
    print(f"  {label}  NDVI={mn:+.4f}  NDWI={mw:+.4f}  NDBI={mb:+.4f}  ({time.time()-t0:.1f}s)")

    del idx, valid
    gc.collect()

df_ts = pd.DataFrame(stats_rows).sort_values("date_frac").reset_index(drop=True)
df_ts.to_csv(f"{OUT_DIR}/stats/index_statistics.csv", index=False)
print(f"\nStats saved. {len(df_ts)} scenes in time-series DataFrame.")
```

### Land Mask

Before computing any area statistics we need to separate land from ocean/lagoon. A pixel counts as permanent water if NDWI > 0.15 in more than 60% of all scenes. Everything else is treated as land.

```python
# ── Land mask ─────────────────────────────────────────────────────────────────
# Problem: every pixel in the scene — ocean, lagoon, everything — is included
# in the vegetation statistics. Kalpitiya is ~14,000 ha; the full scene tile
# is millions of hectares of open water.
#
# Fix: a pixel is "permanent water" if it is water (NDWI > 0.15) in more than
# 60% of all scenes. Everything else is treated as land. We apply this mask
# before computing any area or change statistics.
#
# This runs fast because NDWI TIFs are already on disk from Pass 2.

print("Building persistent land mask from all NDWI TIFs ...\n")

water_count = None    # will hold cumulative count of water observations
total_valid  = None   # will hold count of scenes with valid data per pixel
ref_profile  = None

ndwi_labels = sorted([l for l in LABELS
                       if os.path.exists(f"{OUT_DIR}/ndwi/{l}_ndwi.tif")])

for label in ndwi_labels:
    with rasterio.open(f"{OUT_DIR}/ndwi/{label}_ndwi.tif") as src:
        ndwi_arr    = src.read(1).astype(np.float32)
        if ref_profile is None:
            ref_profile = src.profile.copy()

    valid = ~np.isnan(ndwi_arr)          # pixel has data this scene
    is_water = (ndwi_arr > 0.15) & valid # NDWI > 0.15 → water

    if water_count is None:
        water_count = np.zeros(ndwi_arr.shape, dtype=np.float32)
        total_valid  = np.zeros(ndwi_arr.shape, dtype=np.float32)

    h_ = min(water_count.shape[0], is_water.shape[0])
    w_ = min(water_count.shape[1], is_water.shape[1])
    water_count[:h_, :w_] += is_water[:h_, :w_].astype(np.float32)
    total_valid[:h_, :w_]  += valid[:h_, :w_].astype(np.float32)
    del ndwi_arr
    gc.collect()

# Fraction of valid scenes where this pixel was water
water_frac = np.where(total_valid > 0, water_count / total_valid, 1.0)

# Land = NOT permanently water (threshold: water in fewer than 60% of scenes)
LAND_MASK = (water_frac < 0.60)    # True = land pixel
print(f"  Total pixels in scene      : {LAND_MASK.size:>12,}")
print(f"  Permanent water pixels     : {(~LAND_MASK).sum():>12,}  "
      f"({100*(~LAND_MASK).mean():.1f}%)")
print(f"  Land pixels (study area)   : {LAND_MASK.sum():>12,}  "
      f"({100*LAND_MASK.mean():.1f}%)")
print(f"  Approx land area           : {LAND_MASK.sum() * PX_HA:>12,.0f} ha")

# Save land mask TIF for reference
land_mask_path = f"{OUT_DIR}/stats/land_mask.tif"
lm_profile = ref_profile.copy()
lm_profile.update(count=1, dtype="uint8", compress="lzw", nodata=255)
with rasterio.open(land_mask_path, "w", **lm_profile) as dst:
    dst.write(LAND_MASK.astype(np.uint8)[np.newaxis])
print(f"\nLand mask saved: {land_mask_path}")

# Visualise
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
axes[0].imshow(water_frac, cmap="Blues", vmin=0, vmax=1)
axes[0].set_title("Water fraction across all scenes\n(dark blue = permanent water)",
                   fontweight="bold")
axes[0].axis("off")

axes[1].imshow(LAND_MASK, cmap="YlGn", vmin=0, vmax=1)
axes[1].set_title("Land mask  (green = land used for analysis)",
                   fontweight="bold")
axes[1].axis("off")

plt.suptitle("Persistent land mask — removes ocean and lagoon from statistics",
             fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig(f"{OUT_DIR}/graphs/land_mask.png", dpi=130, bbox_inches="tight")
plt.show()
del water_count, total_valid, water_frac
gc.collect()
```

```python
def pseudo_labels(ndvi_f, ndwi_f, ndbi_f):
    """
    Rule-based pseudo-label generator.
    Only labels pixels we are confident about — ambiguous ones stay -1.

    Classes:
      0 = Water        1 = Vegetation
      2 = Bare Soil    3 = Built-Up / Scrub
    """
    lab = np.full(len(ndvi_f), -1, dtype=np.int8)
    lab[(ndvi_f > 0.05)  & (ndvi_f < 0.22) & (np.abs(ndbi_f) < 0.12)] = 2  # bare
    lab[(ndbi_f > 0.08)  & (ndvi_f < 0.12)]                             = 3  # built-up
    lab[(ndwi_f > 0.15)  & (ndvi_f < 0.02)]                             = 0  # water
    lab[(ndvi_f > 0.32)  & (ndwi_f < -0.05)]                            = 1  # vegetation
    return lab


def build_feature_matrix(stack, ndvi, ndwi, ndbi, nbr, evi):
    """
    Stack bands + indices into (N_valid_pixels, 11) feature matrix.
    Returns (X, valid_mask) where valid_mask is bool (H, W).
    """
    layers = [
        stack[1], stack[2], stack[3],     # B2 B3 B4
        stack[4], stack[5], stack[6],     # B5 B6 B7
        ndvi, ndwi, ndbi, nbr, evi,
    ]
    cube       = np.stack(layers, axis=0)            # (11, H, W)
    valid_mask = ~np.any(np.isnan(cube), axis=0)     # (H, W)
    X = cube.reshape(11, -1).T[valid_mask.ravel()]
    del cube
    return X.astype(np.float32), valid_mask


FEATURE_NAMES = ["B2","B3","B4","B5","B6","B7","NDVI","NDWI","NDBI","NBR","EVI"]
print("Classification helpers defined.")
```

### Model Training — Cross-Scene Accuracy

The previous version showed 100% accuracy because it trained and tested on pseudo-labels from the **same scene**. This version trains on scene 1 and validates on three completely different dates — giving a realistic cross-scene generalisation accuracy.

```python
# ── Pass 3 — Train Random Forest (honest accuracy version) ────────────────────
#
# The previous version got 100% accuracy because:
#   - Labels were generated using NDVI/NDWI/NDBI thresholds
#   - Those same indices were also given as features to the model
#   - The model just re-learned the thresholds it was given — not real learning
#
# Fix: train on scene 1 (2016_02), then validate on THREE different scenes
# that the model has never seen. Cross-scene accuracy is a genuine test of
# whether the spectral patterns generalise across dates and conditions.

ref_label = LABELS[0]
# Use scenes 5, 10, 15 as independent validation (spread across the time series)
val_indices = [4, 9, 14]
val_labels  = [LABELS[i] for i in val_indices if i < len(LABELS)]
print(f"Training scene  : {ref_label}")
print(f"Validation scenes: {val_labels}\n")

# ── Load training scene ───────────────────────────────────────────────────────
with rasterio.open(f"{OUT_DIR}/stacked/{ref_label}_stack.tif") as src:
    ref_stack = src.read().astype(np.float32)

r_ndvi = read_index_tif(ref_label, "ndvi")
r_ndwi = read_index_tif(ref_label, "ndwi")
r_ndbi = read_index_tif(ref_label, "ndbi")
r_nbr  = read_index_tif(ref_label, "nbr")
r_evi  = read_index_tif(ref_label, "evi")

X_all, vmask = build_feature_matrix(ref_stack, r_ndvi, r_ndwi, r_ndbi, r_nbr, r_evi)
del ref_stack, r_nbr, r_evi
gc.collect()

labs_all = pseudo_labels(r_ndvi[vmask], r_ndwi[vmask], r_ndbi[vmask])
del r_ndvi, r_ndwi, r_ndbi, vmask
gc.collect()

keep  = labs_all >= 0
X_lab = X_all[keep]
y_lab = labs_all[keep]
del X_all, labs_all
gc.collect()

print(f"Training pseudo-labels: {len(y_lab):,}")
for cid, cname in CLASS_NAMES.items():
    n = int((y_lab == cid).sum())
    print(f"  {cname:<22} {n:>9,}  ({100*n/len(y_lab):.1f}%)")

# ── Scale ─────────────────────────────────────────────────────────────────────
SCALER   = StandardScaler()
X_sc     = SCALER.fit_transform(X_lab)
del X_lab
gc.collect()

# 75/25 split within the training scene (used only for OOB context)
X_tr, X_te_same, y_tr, y_te_same = train_test_split(
    X_sc, y_lab, test_size=0.25, stratify=y_lab, random_state=42
)
del X_sc, y_lab
gc.collect()

print(f"\nTrain samples : {len(X_tr):,}")
print(f"Training Random Forest (200 trees, all CPU cores) ...")

t0 = time.time()
RF = RandomForestClassifier(
    n_estimators=200, max_depth=15, min_samples_leaf=5,
    max_features="sqrt", class_weight="balanced",
    oob_score=True, n_jobs=-1, random_state=42,
)
RF.fit(X_tr, y_tr)

# ── In-scene test (same scene, held-out pixels) ───────────────────────────────
y_pred_same = RF.predict(X_te_same)
acc_same    = accuracy_score(y_te_same, y_pred_same)
from sklearn.metrics import cohen_kappa_score
kappa_same  = cohen_kappa_score(y_te_same, y_pred_same)

print(f"\nTraining time   : {time.time()-t0:.1f}s")
print(f"OOB score       : {RF.oob_score_:.4f}")
print(f"\n--- In-scene accuracy (same scene holdout) ---")
print(f"Accuracy        : {acc_same:.4f}  ← inflated; model saw this scene's radiometry")
print(f"Cohen Kappa     : {kappa_same:.4f}")
print("NOTE: In-scene accuracy is always near 1.0 with pseudo-labels. See cross-scene below.")

# ── Cross-scene validation (the honest number) ────────────────────────────────
print(f"\n--- Cross-scene accuracy (different scenes — the real test) ---")
cross_accs   = []
cross_kappas = []

for vl in val_labels:
    vstk_path = f"{OUT_DIR}/stacked/{vl}_stack.tif"
    if not os.path.exists(vstk_path):
        print(f"  {vl}: stacked TIF not found — skipping")
        continue

    with rasterio.open(vstk_path) as src:
        vstk = src.read().astype(np.float32)

    v_ndvi = read_index_tif(vl, "ndvi")
    v_ndwi = read_index_tif(vl, "ndwi")
    v_ndbi = read_index_tif(vl, "ndbi")
    v_nbr  = read_index_tif(vl, "nbr")
    v_evi  = read_index_tif(vl, "evi")

    X_v, vm_v = build_feature_matrix(vstk, v_ndvi, v_ndwi, v_ndbi, v_nbr, v_evi)
    del vstk, v_nbr, v_evi
    gc.collect()

    labs_v = pseudo_labels(v_ndvi[vm_v], v_ndwi[vm_v], v_ndbi[vm_v])
    del v_ndvi, v_ndwi, v_ndbi, vm_v
    gc.collect()

    keep_v  = labs_v >= 0
    X_v_lab = X_v[keep_v]
    y_v_lab = labs_v[keep_v]
    del X_v, labs_v
    gc.collect()

    if len(y_v_lab) < 100:
        print(f"  {vl}: too few labelled pixels to validate")
        del X_v_lab, y_v_lab; gc.collect()
        continue

    X_v_sc = SCALER.transform(X_v_lab)
    del X_v_lab
    gc.collect()

    y_v_pred  = RF.predict(X_v_sc)
    del X_v_sc
    gc.collect()

    acc_v   = accuracy_score(y_v_lab, y_v_pred)
    kappa_v = cohen_kappa_score(y_v_lab, y_v_pred)
    cross_accs.append(acc_v)
    cross_kappas.append(kappa_v)

    print(f"  {vl}  accuracy={acc_v:.4f}  kappa={kappa_v:.4f}  "
          f"(n={len(y_v_lab):,})")
    del y_v_lab, y_v_pred

mean_acc   = float(np.mean(cross_accs))   if cross_accs   else 0.0
mean_kappa = float(np.mean(cross_kappas)) if cross_kappas else 0.0

print(f"\n  Mean cross-scene accuracy : {mean_acc:.4f}")
print(f"  Mean cross-scene kappa    : {mean_kappa:.4f}")
print()
if mean_kappa >= 0.80:
    interpret = "Excellent agreement — model generalises well across dates."
elif mean_kappa >= 0.60:
    interpret = "Good agreement — acceptable for land cover mapping."
elif mean_kappa >= 0.40:
    interpret = "Moderate agreement — usable but treat class boundaries cautiously."
else:
    interpret = "Weak agreement — spectral variability across dates is high."
print(f"  Kappa interpretation: {interpret}")

# Store for the evaluation cell below
y_pred_te = y_pred_same
y_te      = y_te_same
CROSS_SCENE_ACCURACY = mean_acc
CROSS_SCENE_KAPPA    = mean_kappa

joblib.dump(RF,     f"{OUT_DIR}/models/rf_model.pkl")
joblib.dump(SCALER, f"{OUT_DIR}/models/scaler.pkl")
del X_tr
gc.collect()
print(f"\nModel saved.")
```

```python
# ── Pass 4 — classify every scene, report land-only area stats ───────────────
# Fix: apply the LAND_MASK when computing class areas so ocean/lagoon pixels
# are excluded. The area numbers will now reflect just the land portion of
# the scene, not the full tile including open water.
#
# If LAND_MASK is not in memory (kernel restart), reload it from disk.

if "LAND_MASK" not in dir():
    print("Loading land mask from disk ...")
    with rasterio.open(f"{OUT_DIR}/stats/land_mask.tif") as src:
        LAND_MASK = src.read(1).astype(bool)

if "RF" not in dir():
    RF     = joblib.load(f"{OUT_DIR}/models/rf_model.pkl")
    SCALER = joblib.load(f"{OUT_DIR}/models/scaler.pkl")
    print("Model loaded from disk.")

print("Classifying all scenes ...\n")
area_records = []

for label in LABELS:
    stk_path  = f"{OUT_DIR}/stacked/{label}_stack.tif"
    cmap_path = f"{OUT_DIR}/classmaps/{label}_classmap.tif"

    if not os.path.exists(stk_path):
        print(f"  {label}: stacked TIF missing")
        continue

    t0 = time.time()

    with rasterio.open(stk_path) as src:
        stk     = src.read().astype(np.float32)
        profile = src.profile.copy()

    H, W = stk.shape[1], stk.shape[2]

    ndvi = read_index_tif(label, "ndvi")
    ndwi = read_index_tif(label, "ndwi")
    ndbi = read_index_tif(label, "ndbi")
    nbr  = read_index_tif(label, "nbr")
    evi  = read_index_tif(label, "evi")

    X_all, vmask = build_feature_matrix(stk, ndvi, ndwi, ndbi, nbr, evi)
    del stk, ndvi, ndwi, ndbi, nbr, evi
    gc.collect()

    X_sc = SCALER.transform(X_all)
    del X_all
    gc.collect()

    CHUNK = 500_000
    preds = np.empty(X_sc.shape[0], dtype=np.int8)
    for start in range(0, len(X_sc), CHUNK):
        preds[start:start+CHUNK] = RF.predict(X_sc[start:start+CHUNK]).astype(np.int8)
    del X_sc
    gc.collect()

    cmap_flat = np.full(H * W, -1, dtype=np.int8)
    cmap_flat[vmask.ravel()] = preds
    cmap = cmap_flat.reshape(H, W)
    del cmap_flat, preds, vmask
    gc.collect()

    # Save classification map
    p1 = profile.copy()
    p1.update(count=1, dtype="float32", compress="lzw", nodata=-1.0, BIGTIFF="YES")
    save_raster(cmap.astype(np.float32), p1, cmap_path)

    # ── Area stats: LAND PIXELS ONLY ──────────────────────────────────────────
    # Align land mask shape to classmap shape (they should match but be safe)
    lm = LAND_MASK[:H, :W]
    land_classified = lm & (cmap >= 0)    # land AND has a valid prediction
    total_land_px   = int(land_classified.sum())

    row = {"label": label, "total_land_px": total_land_px,
           "total_land_ha": round(total_land_px * PX_HA, 1)}
    for cid, cname in CLASS_NAMES.items():
        key = cname.replace(" ", "_").replace("/", "").replace("-", "")
        n   = int(((cmap == cid) & land_classified).sum())
        row[f"{key}_px"]  = n
        row[f"{key}_pct"] = round(100 * n / total_land_px if total_land_px > 0 else 0, 2)
        row[f"{key}_ha"]  = round(n * PX_HA, 1)
    area_records.append(row)

    veg = row.get("Vegetation_pct", 0)
    wat = row.get("Water_pct", 0)
    lha = row.get("total_land_ha", 0)
    print(f"  {label}  veg={veg:.1f}%  water={wat:.1f}%  "
          f"land area={lha:,.0f}ha  ({time.time()-t0:.1f}s)")

    del cmap, lm, land_classified
    gc.collect()

df_areas = pd.DataFrame(area_records)
df_areas.to_csv(f"{OUT_DIR}/stats/class_areas.csv", index=False)
print(f"\nClassification done. Land-only area stats saved.")
print(f"Average land area across scenes: {df_areas['total_land_ha'].mean():,.0f} ha")
```

```python
# ── Land-masked vegetation change stats ──────────────────────────────────────
# Replaces the original veg_change_stats function.
# Now applies LAND_MASK so ocean / lagoon pixels are excluded from
# every loss, gain and area calculation.

def veg_change_stats(ndvi_t1, ndvi_t2, land_mask=None, thresh=VEG_THRESHOLD):
    """
    Compare vegetation cover between two NDVI images.

    If land_mask is provided (bool array, True = land), ocean and lagoon
    pixels are excluded before counting. This makes the ha figures realistic.
    """
    h = min(ndvi_t1.shape[0], ndvi_t2.shape[0])
    w = min(ndvi_t1.shape[1], ndvi_t2.shape[1])
    t1, t2 = ndvi_t1[:h, :w], ndvi_t2[:h, :w]

    both_valid = (~np.isnan(t1)) & (~np.isnan(t2))

    if land_mask is not None:
        lm         = land_mask[:h, :w]
        both_valid = both_valid & lm

    v1 = (t1 > thresh) & both_valid
    v2 = (t2 > thresh) & both_valid
    n1, n2  = int(v1.sum()), int(v2.sum())
    lost    = int((v1 & ~v2).sum())
    gained  = int((~v1 & v2).sum())

    return {
        "n_veg_t1":  n1,  "n_veg_t2":  n2,
        "lost_px":   lost, "gained_px": gained,
        "lost_ha":   round(lost        * PX_HA, 1),
        "gained_ha": round(gained      * PX_HA, 1),
        "net_ha":    round((n2 - n1)   * PX_HA, 1),
        "pct_loss":  round(100 * lost   / n1 if n1 > 0 else 0.0, 2),
        "pct_gain":  round(100 * gained / n1 if n1 > 0 else 0.0, 2),
        "pct_net":   round(100 * (n2-n1) / n1 if n1 > 0 else 0.0, 2),
    }


print("Updated veg_change_stats function loaded (land-mask aware).")
```

```python
# ── Main change detection — land pixels only ──────────────────────────────────
# Load LAND_MASK if not in memory from a previous cell.

if "LAND_MASK" not in dir():
    print("Loading land mask from disk ...")
    with rasterio.open(f"{OUT_DIR}/stats/land_mask.tif") as src:
        LAND_MASK = src.read(1).astype(bool)

sorted_labels = df_ts["label"].tolist()
# Find the latest scene (T2) and then find the earliest scene (T1) with the EXACT SAME MONTH
T2 = sorted_labels[-1]
target_month = T2.split("_")[1]
T1 = sorted_labels[0]  # Fallback
for label in sorted_labels:
    if label.split("_")[1] == target_month and label != T2:
        T1 = label
        break
print(f"Selected {T1} and {T2} for seasonal-adjusted change detection.")

with rasterio.open(f"{OUT_DIR}/ndvi/{T1}_ndvi.tif") as src:
    ndvi_t1  = src.read(1).astype(np.float32)
    ref_prof = src.profile.copy()

with rasterio.open(f"{OUT_DIR}/ndvi/{T2}_ndvi.tif") as src:
    ndvi_t2 = src.read(1).astype(np.float32)

h = min(ndvi_t1.shape[0], ndvi_t2.shape[0])
w = min(ndvi_t1.shape[1], ndvi_t2.shape[1])

# Apply land mask to the difference — ocean pixels become NaN
lm   = LAND_MASK[:h, :w]
diff = (ndvi_t2[:h, :w] - ndvi_t1[:h, :w]).astype(np.float32)
diff[~lm] = np.nan     # hide water pixels from the heatmap

chg = np.zeros_like(diff, dtype=np.int8)
chg[diff < -0.08]    = -1    # loss
chg[diff >  0.08]    =  1    # gain
chg[np.isnan(diff)]  = -99   # no data or water

# Use the land-mask-aware version of the function
vstats = veg_change_stats(ndvi_t1, ndvi_t2, land_mask=LAND_MASK)

print(f"{'='*55}")
print(f"  Vegetation change  {T1}  to  {T2}  (LAND ONLY)")
print(f"{'='*55}")
print(f"  Land veg area at {T1}  : {vstats['n_veg_t1']:>10,} px "
      f"  ({vstats['n_veg_t1'] * PX_HA:>10,.1f} ha)")
print(f"  Land veg area at {T2}  : {vstats['n_veg_t2']:>10,} px "
      f"  ({vstats['n_veg_t2'] * PX_HA:>10,.1f} ha)")
print(f"  Area lost          : {vstats['lost_px']:>10,} px "
      f"  ({vstats['lost_ha']:>8,.1f} ha)  {vstats['pct_loss']:+.2f}%")
print(f"  Area gained        : {vstats['gained_px']:>10,} px "
      f"  ({vstats['gained_ha']:>8,.1f} ha)  {vstats['pct_gain']:+.2f}%")
print(f"  Net                :              "
      f"  ({vstats['net_ha']:>+8,.1f} ha)  {vstats['pct_net']:+.2f}%")
print(f"  Mean land NDVI Δ   : {float(np.nanmean(diff)):>+.4f}")
print(f"{'='*55}")

with open(f"{OUT_DIR}/stats/vegetation_change.json", "w") as f:
    json.dump({**vstats, "T1": T1, "T2": T2,
               "mean_delta": round(float(np.nanmean(diff)), 4),
               "land_only": True}, f, indent=2)

save_raster(diff, ref_prof,
            f"{OUT_DIR}/change/{T1}_to_{T2}_delta_ndvi.tif")
save_raster(chg.astype(np.float32), ref_prof,
            f"{OUT_DIR}/change/{T1}_to_{T2}_change_class.tif")
```

```python
# Transition matrix: which land cover classes converted into which others.

def load_classmap_arr(label):
    path = f"{OUT_DIR}/classmaps/{label}_classmap.tif"
    if not os.path.exists(path):
        return None
    with rasterio.open(path) as src:
        return src.read(1).astype(np.float32)


cm_t1 = load_classmap_arr(T1)
cm_t2 = load_classmap_arr(T2)

if cm_t1 is not None and cm_t2 is not None:
    h_ = min(cm_t1.shape[0], cm_t2.shape[0])
    w_ = min(cm_t1.shape[1], cm_t2.shape[1])
    f1 = cm_t1[:h_, :w_].ravel().astype(int)
    f2 = cm_t2[:h_, :w_].ravel().astype(int)
    valid = (f1 >= 0) & (f2 >= 0)
    mat   = np.zeros((4, 4), dtype=np.int64)
    for i in range(4):
        for j in range(4):
            mat[i, j] = int(((f1 == i) & (f2 == j) & valid).sum())

    cls_labels = [CLASS_NAMES[i] for i in range(4)]
    df_tm = pd.DataFrame(mat, index=cls_labels, columns=cls_labels)
    df_tm.index.name   = f"From ({T1})"
    df_tm.columns.name = f"To ({T2})"
    print(f"Transition matrix ({T1} to {T2}):")
    print(df_tm.to_string())
    df_tm.to_csv(f"{OUT_DIR}/stats/transition_matrix.csv")

    row_sums = mat.sum(axis=1, keepdims=True)
    mat_norm = np.where(row_sums > 0, mat / row_sums.astype(float), 0.0)

    fig, ax = plt.subplots(figsize=(7, 5))
    sns.heatmap(mat_norm, annot=mat, fmt=",d", cmap="YlOrRd",
                xticklabels=cls_labels, yticklabels=cls_labels, ax=ax, linewidths=0.5)
    ax.set_title(f"Class transitions: {T1} to {T2}\n(colour = row fraction, numbers = pixel count)",
                 fontweight="bold")
    ax.set_xlabel(f"To ({T2})")
    ax.set_ylabel(f"From ({T1})")
    plt.tight_layout()
    plt.savefig(f"{OUT_DIR}/change/transition_matrix.png", dpi=130, bbox_inches="tight")
    plt.show()

    del cm_t1, cm_t2, f1, f2, mat, mat_norm
    gc.collect()
else:
    print("Classmap TIFs not found for one or both dates — run classification first.")
```

```python
# ── Sequential pair change detection — land pixels only ───────────────────────

if "LAND_MASK" not in dir():
    with rasterio.open(f"{OUT_DIR}/stats/land_mask.tif") as src:
        LAND_MASK = src.read(1).astype(bool)

pair_rows  = []
diff_stack = []

for i in range(len(sorted_labels) - 1):
    l1, l2 = sorted_labels[i], sorted_labels[i+1]
    p1 = f"{OUT_DIR}/ndvi/{l1}_ndvi.tif"
    p2 = f"{OUT_DIR}/ndvi/{l2}_ndvi.tif"
    if not (os.path.exists(p1) and os.path.exists(p2)):
        continue

    with rasterio.open(p1) as src: n1 = src.read(1).astype(np.float32)
    with rasterio.open(p2) as src: n2 = src.read(1).astype(np.float32)

    h_ = min(n1.shape[0], n2.shape[0])
    w_ = min(n1.shape[1], n2.shape[1])
    d  = (n2[:h_, :w_] - n1[:h_, :w_]).astype(np.float32)
    lm = LAND_MASK[:h_, :w_]
    d[~lm] = np.nan     # mask ocean from diff

    diff_stack.append(d.copy())
    vs = veg_change_stats(n1, n2, land_mask=LAND_MASK)
    pair_rows.append({
        "from":       l1, "to": l2,
        "mean_delta": round(float(np.nanmean(d)), 5),
        **vs
    })
    print(f"  {l1} to {l2}   delta={np.nanmean(d):+.4f}   "
          f"lost={vs['lost_ha']:.0f}ha   gained={vs['gained_ha']:.0f}ha  (land only)")

    del n1, n2, d, lm
    gc.collect()

df_pairs = pd.DataFrame(pair_rows)
df_pairs.to_csv(f"{OUT_DIR}/stats/sequential_change.csv", index=False)

# Summary bar chart
fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
plabels = [str(r["from"])[-5:] for _, r in df_pairs.iterrows()]
bar_c   = ["#cb181d" if v < 0 else "#238b45"
           for v in df_pairs["mean_delta"].to_numpy().astype(float)]

axes[0].bar(range(len(plabels)),
            df_pairs["mean_delta"].to_numpy().astype(float),
            color=bar_c, edgecolor="white")
axes[0].axhline(0, color="black", lw=0.8)
axes[0].set_xticks(range(len(plabels)))
axes[0].set_xticklabels(plabels, rotation=40, ha="right", fontsize=8)
axes[0].set_ylabel("Mean delta NDVI (land only)")
axes[0].set_title("NDVI change per period (land only)", fontweight="bold")

x = np.arange(len(plabels)); w = 0.38
axes[1].bar(x - w/2, df_pairs["lost_ha"].to_numpy().astype(float),
            width=w, color="#cb181d", label="Lost (ha)")
axes[1].bar(x + w/2, df_pairs["gained_ha"].to_numpy().astype(float),
            width=w, color="#238b45", label="Gained (ha)")
axes[1].set_xticks(x)
axes[1].set_xticklabels(plabels, rotation=40, ha="right", fontsize=8)
axes[1].set_ylabel("Area (ha)")
axes[1].set_title("Vegetation area change per period (land only)", fontweight="bold")
axes[1].legend()

plt.tight_layout()
plt.savefig(f"{OUT_DIR}/change/sequential_change.png", dpi=130, bbox_inches="tight")
plt.show()
```

### Land-Only Time Series

Rebuild the NDVI statistics using only land pixels (ocean masked). Scenes where cloud covers more than 50% of the land area are flagged as unreliable and excluded from the trend fitting.

```python
# ── Land-only NDVI time-series ────────────────────────────────────────────────
# Rebuilds df_ts using only land pixels (LAND_MASK applied).
# Also flags scenes where cloud covers more than 50% of the land area —
# those dates are marked unreliable and shown differently in the plot.
#
# Run this cell to replace the df_ts built during the index pass.

if "LAND_MASK" not in dir():
    print("Loading land mask from disk ...")
    with rasterio.open(f"{OUT_DIR}/stats/land_mask.tif") as src:
        LAND_MASK = src.read(1).astype(bool)

print("Rebuilding time-series from land pixels only ...\n")
ts_rows = []

for label in LABELS:
    ndvi_path = f"{OUT_DIR}/ndvi/{label}_ndvi.tif"
    stk_path  = f"{OUT_DIR}/stacked/{label}_stack.tif"
    if not os.path.exists(ndvi_path):
        continue

    with rasterio.open(ndvi_path) as src:
        ndvi_arr = src.read(1).astype(np.float32)

    h = min(ndvi_arr.shape[0], LAND_MASK.shape[0])
    w = min(ndvi_arr.shape[1], LAND_MASK.shape[1])

    lm       = LAND_MASK[:h, :w]
    land_ndvi = ndvi_arr[:h, :w].copy()
    land_ndvi[~lm] = np.nan     # mask out ocean

    valid = land_ndvi[~np.isnan(land_ndvi)].ravel()

    # Cloud fraction over land = (land pixels - valid land pixels) / land pixels
    n_land      = int(lm.sum())
    n_valid     = int(len(valid))
    land_cloud  = 100.0 * (n_land - n_valid) / n_land if n_land > 0 else 100.0

    del ndvi_arr, land_ndvi, lm
    gc.collect()

    if len(valid) < 1000:
        print(f"  {label}: only {len(valid)} valid land pixels — skipping")
        continue

    yr, mo = int(label[:4]), int(label[5:])
    ts_rows.append({
        "label":         label,
        "year":          yr,
        "month":         mo,
        "date_frac":     yr + (mo - 1) / 12.0,
        "mean_ndvi":     float(valid.mean()),
        "std_ndvi":      float(valid.std()),
        "median_ndvi":   float(np.median(valid)),
        "p10":           float(np.percentile(valid, 10)),
        "p90":           float(np.percentile(valid, 90)),
        "pct_dense":     float(100.0 * (valid > 0.50).mean()),
        "pct_moderate":  float(100.0 * ((valid > 0.25) & (valid <= 0.50)).mean()),
        "pct_sparse":    float(100.0 * ((valid > 0.00) & (valid <= 0.25)).mean()),
        "pct_nonveg":    float(100.0 * (valid <= 0.00).mean()),
        "n_valid_land":  n_valid,
        "n_total_land":  n_land,
        "land_cloud_pct": round(land_cloud, 1),
        "reliable":      land_cloud < 50.0,   # True = usable date
    })
    status = "OK  " if land_cloud < 50.0 else "HIGH CLOUD"
    print(f"  {label}  land_cloud={land_cloud:4.1f}%  "
          f"mean_NDVI={valid.mean():+.4f}  [{status}]")
    del valid

df_ts = pd.DataFrame(ts_rows).sort_values("date_frac").reset_index(drop=True)
df_ts.to_csv(f"{OUT_DIR}/stats/ndvi_timeseries_land.csv", index=False)

reliable_n = int(df_ts["reliable"].sum())
print(f"\nTotal scenes   : {len(df_ts)}")
print(f"Reliable (<50% land cloud): {reliable_n}")
print(f"High-cloud scenes excluded from trend: {len(df_ts) - reliable_n}")
```

```python
# ── Regression on reliable scenes only ───────────────────────────────────────
# Use only scenes where land cloud cover is below 50%.
# High-cloud dates are still shown on the plot but excluded from trend fitting.

df_reliable = df_ts[df_ts["reliable"]].copy().reset_index(drop=True)
df_highcld  = df_ts[~df_ts["reliable"]].copy().reset_index(drop=True)

print(f"Scenes used for trend fitting: {len(df_reliable)}")
print(f"High-cloud scenes (shown but not fitted): {len(df_highcld)}\n")

if len(df_reliable) < 3:
    print("WARNING: fewer than 3 reliable scenes. Trend will not be meaningful.")
    print("Consider lowering DOWNSAMPLE or using Sentinel-2 data.")
    # Fall back to all scenes
    df_reliable = df_ts.copy()

X = df_reliable["date_frac"].to_numpy().astype(float).reshape(-1, 1)
y = df_reliable["mean_ndvi"].to_numpy().astype(float)

lr   = LinearRegression().fit(X, y)
poly = make_pipeline(PolynomialFeatures(2), LinearRegression()).fit(X, y)
gbr  = GradientBoostingRegressor(
    n_estimators=100, max_depth=2,
    learning_rate=0.08, subsample=0.8, random_state=42
).fit(X.ravel().reshape(-1, 1), y)

slope, intercept, r_val, p_val, _ = scipy_stats.linregress(X.ravel(), y)
direction = "declining" if slope < 0 else "increasing"
sig_note  = (f"p={p_val:.4f}  "
             f"({'significant' if p_val < 0.05 else 'not significant at 5%'})")

print(f"Linear slope  : {slope:+.6f} NDVI per year  (land pixels only)")
print(f"R squared     : {r_val**2:.4f}")
print(f"Significance  : {sig_note}")
print(f"Direction     : {direction.upper()}")
print()
for name, pred in [("Linear",     lr.predict(X)),
                    ("Polynomial", poly.predict(X)),
                    ("GBR",        gbr.predict(X.ravel().reshape(-1, 1)))]:
    from sklearn.metrics import r2_score, mean_squared_error
    r2   = r2_score(y, pred)
    rmse = float(np.sqrt(mean_squared_error(y, pred)))
    print(f"  {name:<12}  R2={r2:.4f}  RMSE={rmse:.5f}")
```

```python
# ── Forecast plot (reliable scenes highlighted) ───────────────────────────────

fore_dates = np.arange(float(df_reliable["date_frac"].max()) + 0.25, 2031.0, 0.25)
X_f        = fore_dates.reshape(-1, 1)
fc_lr      = np.clip(lr.predict(X_f),                         0.0, 0.9)
fc_poly    = np.clip(poly.predict(X_f),                       0.0, 0.9)
fc_gbr     = np.clip(gbr.predict(X_f.ravel().reshape(-1, 1)), 0.0, 0.9)
fc_ens     = (fc_lr + fc_poly + fc_gbr) / 3.0

df_fore = pd.DataFrame({
    "date_frac":  fore_dates,
    "ensemble":   fc_ens,
    "linear":     fc_lr,
    "polynomial": fc_poly,
    "gbr":        fc_gbr,
})
df_fore.to_csv(f"{OUT_DIR}/stats/ndvi_forecast_2030.csv", index=False)
df_ts.to_csv(f"{OUT_DIR}/stats/ndvi_timeseries.csv", index=False)

fig, ax = plt.subplots(figsize=(14, 6))

# High-cloud scenes (unreliable) shown as faint grey dots
if len(df_highcld) > 0:
    ax.scatter(df_highcld["date_frac"].to_numpy().astype(float),
               df_highcld["mean_ndvi"].to_numpy().astype(float),
               color="#bbbbbb", s=60, zorder=3,
               label="High-cloud scene (excluded from trend)")
    for _, r in df_highcld.iterrows():
        ax.annotate(f"{float(r['mean_ndvi']):.3f}\n({r['land_cloud_pct']:.0f}% cld)",
                    xy=(float(r["date_frac"]), float(r["mean_ndvi"])),
                    xytext=(3, 8), textcoords="offset points",
                    fontsize=7, color="#999999")

# Reliable scenes with uncertainty band
x_rel = df_reliable["date_frac"].to_numpy().astype(float)
y_rel = df_reliable["mean_ndvi"].to_numpy().astype(float)
s_rel = df_reliable["std_ndvi"].to_numpy().astype(float)
p10_r = df_reliable["p10"].to_numpy().astype(float)
p90_r = df_reliable["p90"].to_numpy().astype(float)

ax.fill_between(x_rel, p10_r, p90_r,
                alpha=0.12, color="#238b45", label="P10-P90 range (land)")
ax.fill_between(x_rel, y_rel - s_rel, y_rel + s_rel,
                alpha=0.25, color="#238b45", label="+-1 SD")
ax.plot(x_rel, y_rel, "o-", color="#1a6b38", lw=2.5, ms=9, zorder=5,
        label="Observed NDVI (land, reliable scenes)")

for i, (xv, yv) in enumerate(zip(x_rel, y_rel)):
    ax.annotate(f"{yv:.3f}",
                xy=(xv, yv), xytext=(3, 8),
                textcoords="offset points", fontsize=8, color="#1a6b38")

# Trend line through the full date range
x_line = np.linspace(float(df_ts["date_frac"].min()), 2031, 300).reshape(-1, 1)
ax.plot(x_line, lr.predict(x_line), "--", color="#aaa", lw=1.5,
        label=f"Linear trend  ({slope:+.5f}/yr,  {sig_note})")

# Individual model forecasts (light)
ax.plot(fore_dates, fc_lr,   "-", color="#74add1", lw=1.3, alpha=0.7, label="Linear fc")
ax.plot(fore_dates, fc_poly, "-", color="#fdae61", lw=1.3, alpha=0.7, label="Poly fc")
ax.plot(fore_dates, fc_gbr,  "-", color="#abd9e9", lw=1.3, alpha=0.7, label="GBR fc")

# Ensemble
fc_ens_arr = fc_ens
ax.fill_between(fore_dates, fc_ens_arr - 0.03, fc_ens_arr + 0.03,
                alpha=0.2, color="#cb181d")
ax.plot(fore_dates, fc_ens_arr, "-", color="#cb181d", lw=2.5,
        label="Ensemble forecast (+-uncertainty)")
ax.axvline(float(df_reliable["date_frac"].max()),
           color="#555", ls=":", lw=2, label="Forecast start")

ax.set_xlabel("Year", fontsize=12)
ax.set_ylabel("Mean NDVI (land pixels only)", fontsize=12)
ax.set_title(
    f"NDVI trend and forecast — Kalpitiya, Sri Lanka (land only, 2016-2030)\n"
    f"Trend: {direction.upper()}  |  slope={slope:+.5f}/yr  |  {sig_note}",
    fontsize=13, fontweight="bold")
ax.legend(loc="upper left", fontsize=8, ncol=2)
ax.set_ylim(-0.10, 0.80)
ax.set_xlim(2015.5, 2031.5)
ax.grid(alpha=0.25)
plt.tight_layout()
plt.savefig(f"{OUT_DIR}/graphs/05_ndvi_forecast.png", dpi=130, bbox_inches="tight")
plt.show()
```
